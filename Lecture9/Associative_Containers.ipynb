## Motivation

The following code simulates the outcomes of rolling a die multiple times.

````python
dice_rolls = [random.randint(1, 6) for i in range(10)]
print(*dice_rolls)
````

**What is the distribution?**

````python
plt.figure(0, clear=True)
distribution = [dice_rolls.count(i) / len(dice_rolls) for i in range(7)]
plt.stem(range(7), distribution)
plt.xlabel("Outcomes")
plt.title("Distribution")
plt.ylim(0, 1)
plt.show()
````

In the above code, `distribution[i]` stores the fractional count of outcome `i`.

However, `distribution[0]` is `0` because a die does not have outcome `0`. Can we avoid such redundancy?

````python
distinct_outcomes = [
    outcome for outcome in range(1, 7) if dice_rolls.count(outcome) > 0
]
distribution = [
    dice_rolls.count(distinct_outcomes[i]) / len(dice_rolls)
    for i in range(len(distinct_outcomes))
]

plt.figure(1, clear=True)
plt.stem(distinct_outcomes, distribution)
plt.xlabel("Outcomes")
plt.title("Distribution")
plt.ylim(0, 1)
plt.show()
````

In the above code, 
- `distinct_outcomes` stores the list of distinct outcomes, and
- `distribution[distinct_outcomes[i]]` stores the fractional count of the `i`-th distinct outcome.

**What about finding the distribution of characters in an article?**

There are over 1 million unicode characters. Can we:

- Obtain the distribution efficiently without creating an entry for each unicode character?
- Avoid keeping a separate set of distinct characters in addition to their distribution?
- Access the fractional count of a particular character efficiently without searching through the list of distinct outcomes?

It is desirable to have a composite data type that
- can keep a set of *unique keys of different types* (such as the characters in our example), and
- associate to different keys possibly different *values of any types* (such as the fractional counts of the characters).

Such a data structure is called an [associative container](https://en.wikipedia.org/wiki/Associative_containers).

### Comprehension

The following function uses a one-line dictionary comprehension to return the distribution of items in a sequence:

````python
def distribute(seq):
    return {k: seq.count(k) / len(seq) for k in set(seq)}
````

````python
def plot_distribution(seq, fig_num=None):
    dist = distribute(seq)
    plt.figure(fig_num, clear=True)
    plt.stem(
        dist.keys(),  # set-like view of the keys
        dist.values(),  # view of the values
    )
    plt.xlabel("Items")
    plt.title("Distribution")
    plt.ylim(0, 1)
    plt.show()


plot_distribution("What is the distribution of different characters?", 2)
````

- The object methods `keys` and `values` provide a dynamic [view](https://docs.python.org/3/glossary.html#term-dictionary-view) of the keys.  
- Unlike a copy, subsequent changes to the dictionary are also reflected in a previously returned view.
- `items` provides a set-like view of the key-value pairs.

````python
%%pytutor
a = dict(enumerate("abc"))
views = a.keys(), a.values(), a.items()
print(a.pop(1))  # remove the key 1 and its associated value
print(a.popitem())  # remove and return one key-value pair
a.clear()  # clear the dictionary
````

`set` has `pop` and `clear` but not `popitem`. However, `set.pop` behaves like `dict.popitem` instead of `dict.pop`.

````python
%%pytutor
a = set("abc")
print(a.pop())  # remove and return an element
a.clear()  # clear the set
````

::::{important} A key for `dict` or `set` must be [hashable](https://docs.python.org/3/glossary.html#term-hashable):

- It has a hash value (returned by the method `__hash__`) that never changes during its lifetime, and
- can be compared (using the method `__eq__`) to other objects. *Hashable objects which are equal must have the same hash value.*
::::

Python also uses dictionaries to implement global/local frames, and [hash collisions can slow down the lookup process](https://stackoverflow.com/questions/8271139/why-is-early-return-slower-than-else).

**What are hashable?**

````python
def hashable(obj):
    try:
        hash(obj)
    except Exception:
        return False
    return True


for i in 0, 0.0, 0j, "", False, (), [], {}, set(), frozenset(), ([],):
    if hashable(i):
        print("{} may be hashable. E.g., hash({!r}) == {}".format(type(i), i, hash(i)))
    else:
        print("{} may not be hashable.".format(type(i)))
````

:class: dropdown

In Python, the hash value of an object must remain stable for as long as the object is used as a key in a hash‑based container such as a `dict` or `set`.  When an object is mutable, its internal state can change after it has been inserted into one of these containers.  If those changes affect the object’s hash, the container’s internal bucket mapping becomes invalid—looking up the key may fail, collisions balloon, or the structure can even corrupt itself.  To avoid these subtle and hard‑to‑diagnose bugs, Python marks most mutable types (lists, dictionaries, bytearrays, etc.) as unhashable by raising a `TypeError` when `hash()` is called.  

The only mutable types that bypass this rule are *some* container types that store only hashable items (like `list` or `dict`) whose own hash is never used directly; instead, they rely on the hashes of their elements internally.  But when a type’s identity itself can change in a way that would alter the hash, the safest design is to make it explicitly unhashable, ensuring that instances can only be used as values or elements in unordered collections, but never as keys in hash‑tables.

::::

````python
%%hermes --no-context
Explain in a paragraph or two why `dict` does not have any
immutable counterpart but `set` has an immutable counterpart
called `frozenset`.
````

::::{seealso}
:class: dropdown

Because a `dict` is *by definition* an association that changes over time.  
Its internal structure must be able to add and remove key‑value pairs, rearrange
bucket locations, and grow or shrink in place.  Implementing an immutable
mapping that still supports the same efficient look‑ups would require a
persistent data‑structure that copies or rebalances the entire table at every
update, which is far more complex and slower than a normal mutable hash table.
Python’s `MappingProxyType` can give a *view* of a read‑only dictionary, but the
actual dictionary it wraps is still mutable.

A `set`, on the other hand, is just a collection of hashable elements with no
values attached.  The immutable form, `frozenset`, can be implemented as a
compact hash of its elements, which can be computed once and stored.  Since
there are no “values” to mutate and the internal representation can be a
simple immutable hash table or bitmap, Python can afford a small, efficient
`frozenset` implementation.  Hence, while `dict` lacks an immutable counterpart
due to the inherent mutability of key–value associations, `set` has `frozenset`
because the underlying data structure remains straightforward to make
immutable.

To avoid `KeyError`, we can check if a key is in a dictionary efficiently (due to hashing) using the `in` operator.  
The following is a different implementation of `distribute`.

````python
def distribute(seq):
    dist = {}
    for i in seq:
        dist[i] = (dist[i] if i in dist else 0) + 1 / len(seq)
    return dist


plot_distribution("What is the distribution of different characters?", 3)
````

| Feature                     | `Set.discard(elem)` | `Set.remove(elem)` | `dict.pop(key[, default])` | `del`  (statement) |
|------------------------------|---------------------|--------------------|---------------------------|-------------------|
| **What it is**               | Method of `set`     | Method of `set`    | Method of `dict`          | Language statement|
| **Applicable containers**   | `set` (and `frozenset` is immutable so no method) | `set` | `dict` | Any mutable sequence or mapping, or a variable name |
| **Effect**                   | Remove element if present, otherwise do nothing | Remove element; must exist | Removes key/value pair and returns value | Removes name binding or container item |
| **Error on missing key**    | **No** – silently ignores | **Yes** – raises `KeyError` | No error if `default` supplied, otherwise raises `KeyError` | Raises `NameError` for undefined variable or `KeyError`/`IndexError` for absent key/index |
| **Return value**            | None | None | Removed value (or `default` if key absent) | None |
| **Typical use‑case**        | “Delete if present” where absence is acceptable | “Delete and guarantee existence” – error reveals logic bug | “Remove key and capture its value” with optional fallback | “Delete a variable, list element, dict entry, etc., when you know it exists or want explicit error” |
| **Granularity**             | Single element | Single element | Single key| Single variable, slice, or container item |
| **Examples**                | `my_set.discard(3)` | `my_set.remove(3)` | `val = my_dict.pop('a', None)` | `del my_dict['a']`
`del my_list[0]`
`del my_var` |

**Bottom line** –  
- Use **`discard`** when you want a safe “remove if present” operation.  
- Use **`remove`** when the element must exist and its absence should raise an error.  
- Use **`del`** for general deletion of variables or indexed/keyed items when you’re certain the target exists or want an explicit exception.

`set` overloads many other operators:

````python
%%pytutor
a, b = {1, 2}, {2, 3}

union = a | b
assert all(i in union for i in a) and all(i in union for i in b)

intersection = a & b
assert all(i in a and i in b for i in intersection)

assert intersection <= a <= union  # subset
assert union > b > intersection  # proper superset
assert len(a) + len(b) == len(intersection) + len(union)

symmetric_difference = a ^ b
assert all((i in a or i in b) and not (i in a and i in b) for i in symmetric_difference)
assert symmetric_difference == union - intersection
assert set.isdisjoint(intersection, symmetric_difference)
assert len(union) == len(intersection) + len(symmetric_difference)
````

The following uses `&` and `-` to compare the sets of public attributes for `set` and `dict`:

````python
set_attributes = {attr for attr in dir(set) if attr[0] != "_"}
dict_attributes = {attr for attr in dir(dict) if attr[0] != "_"}
print("Common attributes:", ", ".join(set_attributes & dict_attributes))
print("dict-specific attributes:", ", ".join(dict_attributes - set_attributes))
print("set-specific attributes:", ", ".join(set_attributes - dict_attributes))
````

:class: dropdown

When to use each operation on a **Python `dict`**

| # | Method / keyword | Applicable to `dict` | What it does | Typical use‑case | Notes |
|---|------------------|----------------------|--------------|-----------------|-------|
| 1 | `del d[key]` | ✅ | Deletes the key‑value pair **without returning anything**. Raises `KeyError` if the key is missing. | • You **know** the key exists and you simply want to remove it.
• You **don’t need** the removed value.
• You want a clear, single‑statement deletion. | • Short‑form of `d.pop(key)` when you don’t care about the value.
• Use the `try … except KeyError:` pattern if the key might be absent and you want a custom response. |
| 2 | `d.pop(key, default)` | ✅ | Removes the key‑value pair and **returns the value**. If `key` is missing, returns `default` (or raises `KeyError` if `default` isn’t supplied). | • You need the value that was stored for a key.
• You want the “remove‑and‑get” operation in one expression.
• You want graceful handling of missing keys via a fallback. | • Avoids a separate lookup (`value = d[key]`) followed by `del d[key]`. |
| 3 | `d.popitem()` | ✅ | Removes and returns **one arbitrary key‑value pair** as a tuple. In CPython 3.7+ it’s the **last inserted** pair (LIFO). Raises `KeyError` if the dictionary is empty. | • Incrementally empty a dictionary (e.g., pop items until empty).
• Implement a LIFO/FIFO queue or LRU cache backed by a dict.
• Consume items while iterating. | • No argument; use only when you don’t care which item is removed (or LIFO semantics are acceptable). |
| 4 | `discard()` | ❌ | *Set* method that removes an element if present; does nothing otherwise. |
| 5 | `remove()` | ❌ | *Set* method that removes an element; raises `KeyError` if absent. |

> **Important**: `discard` and `remove` exist on the `set` type, not on `dict`. If you try to call them on a dictionary you’ll get a `AttributeError`.  
> For dictionaries, the only valid “remove” utilities are:  
> – `del` (no return)  
> – `pop` (return value, optional default)  
> – `popitem` (no key specified, returns an arbitrary pair)

---

Quick reference

| Scenario | Choose | Reason |
|----------|--------|--------|
| Key definitely exists, no value needed | `del d[key]` | Clean, explicit deletion |
| Key definitely exists, you need the value | `d.pop(key)` | Remove and retrieve in one step |
| Key might be missing; you want a safe default | `d.pop(key, default)` | No exception, you get `default` |
| You want to empty or consume a dict incrementally | `d.popitem()` | Pops arbitrary pair (LIFO since 3.7) |
| **Set** removal with no error on missing | `s.discard(item)` | *Not a dict* |
| **Set** removal that must exist | `s.remove(item)` | *Not a dict* |

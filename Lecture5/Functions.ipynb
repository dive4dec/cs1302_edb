## What is a Function?

A function is a *callable* object, e.g.:

````python
callable(callable), callable(1)
````

The function `callable` is callable because
- it can be *called/invoked* with some input *arguments/parameters* such as `1` enclosed by parentheses `()`, and then
- *returns* some value computed from the input arguments, such as the boolean value `False` to indicate that the input argument `1` is not callable.

A function can be defined using the [`def` keyword](https://docs.python.org/3/reference/compound_stmts.html#function-definitions)[^def_tutorial]:

```ebnf
[decorators] "def" funcname [type_params] "(" [parameter_list] ")"
                           ["->" expression] ":" suite
```

[^def_tutorial]: The [tutorial on `def`](https://docs.python.org/3/tutorial/controlflow.html#defining-functions) may be easier to read than the formal definition.

E.g., a simple function that prints "Hello, World!" can be defined as follows:

````python
# Function definition
def say_hello():
    print("Hello, World!")
````

````python
# Function invocation
say_hello()
````

To make a function more powerful and solve different problems,
- use a [return statement](https://docs.python.org/3/reference/simple_stmts.html#the-return-statement) to return a value that
- depends on some input arguments.

````python
def increment(x):
    return x + 1


increment(3)
````

A function must have a return value. By default `None` is returned if the function does not have a return statement:

````python
print(f"The return value is {say_hello()}.", )
````

We can also have multiple input arguments.

````python
def length_of_hypotenuse(a, b):
    return (a ** 2 + b ** 2) ** 0.5

length_of_hypotenuse(1, 2), length_of_hypotenuse(3, 4)
````

The arguments are evaluated from left to right:

````python
print("1st input:", input(), "\n2nd input:", input())
````

::::{important}

In summary, a function in Python
- may have any number (even zero) of input arguments, evaluated from left to right,
- may have *side effects* such as printing something to the standard output, and
- must have a return value, which can be `None`.

::::{seealso}
:class: dropdown

Here's a breakdown of the three parameter-passing mechanisms:

1. Call by Value (C)

- A **copy** of the argument's value is passed to the function.
- Modifications to the parameter inside the function **do not affect** the original variable.
- For structs, the entire struct is copied (no reference semantics for normal parameters).
- Pointers are also passed by value — the pointer itself is copied, so reassigning it locally doesn't affect the caller's pointer, though dereferencing can mutate the pointed-to data.

```c
void f(int x) { x = 10; }   // caller's x unchanged
```

2. Call by Reference (C-style "reference semantics" via pointers; C++ `&` references)

C itself doesn't have true call-by-reference, but the effect is achieved with **pointers**:

- The function receives a **memory address** (a copy of the pointer), so it can **mutate the original variable** through dereferencing.
- C++ added true reference parameters (`void f(int &x)`) where aliasing is implicit — the parameter **is** the original variable.

```c
void f(int *x) { *x = 10; }   // caller's variable is modified
```

3. Call by Object Reference / Call by Sharing (Python & Java)

This is the model used by both Python and Java, sometimes called **call by sharing**:

- What's passed is a **reference (pointer) to an object** — but the reference itself is passed **by value** (i.e., copied).
- **Mutating** the object inside the function **is visible** to the caller because both references point to the same object.
- **Reassigning** the parameter (`x = something`) only changes the local copy of the reference — the caller sees **no change**.

```python
# Python
def f(lst):
    lst.append(1)   # caller sees this — same object mutated
    lst = []         # caller does NOT see this — local rebind only

my_list = [0]
f(my_list)  # my_list is now [0, 1]
```

```java
// Java — same behaviour
void f(List<Integer> lst) {
    lst.add(1);     // visible to caller
    lst = new ArrayList<>();  // NOT visible to caller
}
```

**Summary table**

| Mechanism | What's passed | Can mutate caller's data? | Can rebind caller's variable? |
|---|---|---|---|
| **Call by value** (C) | Copy of the value | No | No |
| **Call by reference** (C++ `&`) | Alias of the variable | Yes | Yes |
| **Call by object reference** (Python, Java) | Copy of the reference | Mutations visible | No — rebind is local only |

The key takeaway: Python and Java sit **between** call-by-value and call-by-reference. You get shared object identity (so mutations propagate), but not shared variable identity (so reassignments don't).

A fundamental property of functions in Python is that they are [*first-class
citizens*](https://en.wikipedia.org/wiki/First-class_function), which means
that a function can be
1. assigned to a variable,
2. passed as an input argument, and
3. returned by a function.

````python
%%hermes --no-context
Are there programming languages that do not treat functions as
first-class citizens? Why?
````

::::{seealso}
:class: dropdown

Several languages do not treat functions as first-class citizens. Notable examples:

C (the classic example)

- Functions **cannot** be stored in variables, returned from other functions, or passed as arguments directly.
- The closest you get is **function pointers**, which are a second-class type — they carry no closure information (no captured environment), have limited type safety, and can't be created dynamically.
- No nested functions, no lambdas.

COBOL / Fortran (traditional versions)

- Functions exist, but cannot be passed, stored, or returned. No function literals, no higher-order functions.

MATLAB (partially)

- Function handles exist, but with limitations compared to first-class treatment.

Why?

Several design reasons explain the absence:

1. **Performance / runtime model** — Storing functions dynamically requires heap allocation, garbage collection, and runtime dispatch. C was designed in the 1970s for minimal overhead and direct mapping to machine code.

2. **No closures** — First-class functions in most languages come paired with closures (capturing surrounding variables). That requires runtime environments to manage. Languages without closures struggle to justify the overhead of full first-class treatment.

3. **Simplicity & compile-time resolution** — When function calls are resolved at compile time and bound statically, the compiler can inline, optimize, and enforce calling conventions. Allowing functions to flow freely through data structures introduces dynamic dispatch and indirection that complicates compilation and optimization.

4. **Memory layout** — In C, functions live in the code segment with fixed addresses. Treating them as data requires them to behave like heap-allocated objects, which clashes with the language's execution model.

5. **Philosophical / procedural orientation** — Languages designed for structured programming (C, Pascal, COBOL) viewed functions as declarative building blocks of programs, not as values that could be manipulated.

::::{seealso}
:class: dropdown

You can create self-recursion in a lambda using a **fixed-point combinator** — the **Z combinator** (Python's version; the Y combinator works for lazy languages).

The trick

A lambda has no name, so it can't call itself directly. The Z combinator passes a "wrapped" version of the function to itself, enabling self-recursion:

```python
Z = lambda f: (lambda x: f(lambda *args: x(x)(*args)))(lambda x: f(lambda *args: x(x)(*args)))
```

Infinite loop

```python
Z(lambda f: lambda: f())()
```

This creates a lambda that calls itself forever. In practice, Python will eventually raise `RecursionError` because of its call-stack limit — but conceptually it is an infinite loop expressed **purely with lambdas**.

How it works (intuition)

- `Z` takes a function `f` that expects to receive its own recursive version as an argument.
- The `(lambda x: ...)(lambda x: ...)` pattern forces **self-application** — the inner lambda calls `x(x)`, which is the mechanism that generates the fixed point.
- `f(lambda *args: x(x)(*args))` wraps the self-application so it only triggers on invocation (avoiding infinite evaluation in an eager language like Python).

### Integer Square Root

To improve the efficiency, consider the following sufficient and necessary condition for perfect squares:

::::{prf:proposition} integer square root
:label: pro:integer-sqrt

An integer $n$ is a perfect square iff

$$
n = (\lfloor \sqrt{n} \rfloor)^2,
$$ (eq:perfect_square)

namely, that the number is the square of its integer square root.

::::

A simple implementation is as follows:

````python
def is_perfect_square(n):
    # check if n is the square of its integer square root
    return n == int(n**0.5) ** 2

assert is_perfect_square(10**10)
````

Note that it fixed the efficiency issue on the test case with `n` being `10**10`. Let's run all the test cases:

````python
test()
````

::::{note} Does it pass all the test cases?
:class: dropdown

`assert is_perfect_square(10**100)` fails because of the finite precision of floating point numbers.

::::

Perhaps we should use `math.isclose` instead of `==`, since floating-point precision loss in `n**0.5` for very large `n`.

````python
def is_perfect_square(n):
    return math.isclose(n, int(n**0.5) ** 2)


assert is_perfect_square(10**100)
````

Note that it can correctly say `10**100` is a perfect square. Let's run all the test cases:

There are [better ways to compute integer square
root](https://en.wikipedia.org/wiki/Integer_square_root). Binary search is a
relatively easy one to try first, although it is not the best choice.

````python
%%hermes --no-context
Explain very briefly how integer square root can be implemented in Python using
binary search.
````

::::{seealso}
:class: dropdown

Binary search works because `x²` is monotonically increasing for non-negative integers. We search for the largest integer `mid` such that `mid² ≤ n`.

```python
def isqrt(n):
    if n < 0: raise ValueError
    lo, hi = 0, n
    while lo <= hi:
        mid = (lo + hi) // 2
        if mid * mid <= n:
            lo = mid + 1
        else:
            hi = mid - 1
    return hi  # hi is the largest int where hi² ≤ n
```

**How it works:**
- If `mid² ≤ n`, the answer is `mid` or higher → move `lo` up.
- If `mid² > n`, the answer is below `mid` → move `hi` down.
- When the loop ends, `hi` holds the integer square root.

````python
%%hermes --no-context
Explain briefly in two paragraph how isqrt is implemented as an 
adaptive-precision pure-integer version of Newton's iteration.
````

::::{seealso}
:class: dropdown

Newton's method finds roots by iterating $x_{k+1} = \frac{1}{2}(x_k + \frac{n}{x_k})$. For integer square root, every division is replaced with integer division (`//`), so the entire computation stays in the integer domain with zero floating-point arithmetic. Starting from an initial guess (like `n` itself), the sequence rapidly converges downward. Because integer division truncates toward zero, the iterates monotonically decrease once they fall below the true root and naturally settle at the correct integer value without overshooting.

The "adaptive precision" comes from Newton's quadratic convergence: each step roughly doubles the number of correct bits. With large integers, this means the algorithm needs only \(O(\log \log n)\) iterations regardless of bit length, while the cost per iteration adapts to the current operand size—early steps are cheap (few bits), and later steps refine the result with larger multiplications and divisions. The loop terminates when \(x_{k+1} \ge x_k\), which is the exact integer square root.

The above uses the wildcard `*` to import ([nearly](https://docs.python.org/3/tutorial/modules.html#more-on-modules)) all the functions/variables provided in `math`.

::::{caution} What if different packages define the same function?
:class: dropdown

In the following code:

- The function `pow` imported from `math` overwrites the built-in function `pow`.  
- Unlike the built-in function, `pow` from `math` returns only floats but not integers or complex numbers. 
- We say that the import statement *polluted the namespace of the global frame* and caused a *name collision*.

::::

````python
%%optlite -h 400
print("{}".format(pow(-1, 2)))
print("{:.2f}".format(pow(-1, 1 / 2)))
from math import *

print("{}".format(pow(-1, 2)))
print("{:.2f}".format(pow(-1, 1 / 2)))
````

To avoid name collisions, it is a good practice to use the full name (*fully-qualified name*) such as `math.pow` prefixed with the module.

````python
import math

print("{:.2f}, {:.2f}".format(math.pow(-1, 2), pow(-1, 1 / 2)))
````

Using the full name can be problematic if the name of a module is very long. There can even be a hierarchical structure.
E.g., to plot a sequence using `pyplot` module from `matplotlib` package:

````python
%matplotlib widget
import matplotlib.pyplot

matplotlib.pyplot.stem([4, 3, 2, 1])
matplotlib.pyplot.ylabel(r"$x_n$")
matplotlib.pyplot.xlabel(r"$n$")
matplotlib.pyplot.title("A sequence of numbers")
matplotlib.pyplot.show()
````

In Python, modules can be structured into [packages](https://docs.python.org/3/tutorial/modules.html#packages), which are themselves modules that can be imported. It is common to rename `matplotlib.pyplot` as `plt`:

````python
import matplotlib.pyplot as plt

plt.stem([4, 3, 2, 1])
plt.ylabel(r"$x_n$")
plt.xlabel(r"$n$")
plt.title("A sequence of numbers")
plt.show()
````

We can also rename a function as we import it to avoid name collision:

````python
from math import pow as fpow

fpow(2, 2), pow(2, 2)
````

## Documentation

Understanding how to properly document a function is crucial for maintaining clear and efficient code. It also allows others to use the code properly to avoid bugs. How should one go about documenting a function effectively? As an example:

````python
# Author: John Doe
# Last modified: 2020-09-14
def increment(x):
    """Increment by 1.

    A simple demo of
    - parameter passing,
    - return statement, and
    - function documentation."""
    return x + 1  # + operation is used and may fail for 'str'
````

The `help` command shows the docstring we write 
- at the beginning of the function body
- delimited using triple single/double quotes.

````python
help(increment)
````

The docstring should contain the *usage guide*, i.e., information for new users to call the function properly. See Python style guide (PEP 257) for
- [one-line docstrings](https://www.python.org/dev/peps/pep-0257/#one-line-docstrings) and
- [multi-line docstrings](https://www.python.org/dev/peps/pep-0257/#multi-line-docstrings).

::::{note} Why doesn't `help` show the comments that start with `#`?
:class: dropdown

```python
# Author: John Doe
# Last modified: 2020-09-14
def increment(x):
    ...
    return x + 1  # + operation is used and may fail for 'str'
```

Those comments are not usage guide. They are intended for programmers who need to maintain/extend the function definition:

- Information about the author and modification date facilitate communications among programmers.
- Comments within the code help explain important and not-so-obvious implementation details.

::::

We can also [annotate](https://docs.python.org/3/library/typing.html) the function with *type hints* to indicate the types of the arguments and return value.

````python
# Author: John Doe
# Last modified: 2020-09-14
def increment(x: float) -> float:
    """Increment by 1.

    A simple demo of
    - parameter passing,
    - return statement, and
    - function documentation."""
    return x + 1  # + operation is used and may fail for 'str'


help(increment)
````

Annotations, if done right, can make the code easier to understand. However, annotations are not enforced by the Python interpreter.[^type-checking]

[^type-checking]: Type checking may be enforced by a supporting editor or packages such as [`pydantic`](https://docs.pydantic.dev/latest/) and [`mypy`](https://mypy.readthedocs.io/en/stable/).

````python
def increment_user_input():
    return increment(input())  # does not raise error even though input returns str
````

Does calling the function lead to any error:

```python
increment_user_input()
```

The types can also be described in the docstring following the
[Numpy](https://numpydoc.readthedocs.io/en/latest/format.html#parameters) or
[Google](https://google.github.io/styleguide/pyguide.html#383-functions-and-methods)
style.

````python
# Author: John Doe
# Last modified: 2020-09-14
def increment(x: float) -> float:
    """Increment by 1.

    A simple demo of
    - parameter passing,
    - return statement, and
    - function documentation.

    Parameters
    ----------
    x: float
        Value to be incremented.

    Returns
    -------
    float:
        Value of x incremented by 1.
    """
    return x + 1  # + operation is used and may fail for 'str'


help(increment)
````

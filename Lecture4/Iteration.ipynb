### While Loop

We can use the [`while` statement](https://docs.python.org/3/reference/compound_stmts.html#while) to repeatedly execute certain code until a specified looping condition is false. A simplified version of the syntax is:

```ebnf
while_stmt ::=  "while" assignment_expression ":" suite
```

````python
%%flowchart
st=>start: Start
cond1=>condition: assignment_expression
suite1=>operation: suite
e=>end

st(right)->cond1
cond1(yes, right)->suite1(right)->cond1
cond1(no)->e
````

In other words, `suite` is repeatedly executed as long as `assignment_expression` is [interpreted as `True` by the control flow statement](https://docs.python.org/3/reference/expressions.html#booleans).

As an example, the following program keeps asking for user input until the input is non-empty.

````python
while not input("Input something please:"):
    pass
````

If user inputs nothing, `input` returns an empty string `''`, which is regarded as `False` by control flow statements, and so the looping condition `not input('...')` is `True`.

Consider the following while loop which attempts to solve [](#ex:print-upto):

````python
num = int(input(">"))
i = -1
while i != num:
    i += 1
    print(i)
````

When writing loops, it's essential to avoid creating unintended *infinite loops*, which can bring your server to a grinding halt.

::::

Indeed, it is impossible to automatically determine whether a program terminates, a challenge known as the [Halting problem](https://en.wikipedia.org/wiki/Halting_problem).

````python
%%hermes --no-context
Prove that it is impossible to automatically determine whether a program terminates.
````

::::{seealso}
:class: dropdown

This is **Turing's Halting Problem** — proved undecidable in 1936. Here's the proof by contradiction:

---

**Step 1: Assume the opposite.** Suppose there exists a program `halts(P, I)` that takes two inputs — a program `P` and an input `I` — and correctly returns `True` if `P(I)` terminates and `False` if `P(I)` runs forever.

**Step 2: Build a paradoxical program.** Using our hypothetical `halts`, construct:

```python
def paradox(P):
    if halts(P, P):
        loop_forever()
    else:
        return  # terminate immediately
```

This program takes another program `P`, feeds it its own code as input (`P(P)`), and does the *opposite* of what `halts` predicts.

**Step 3: Feed it to itself.** What happens when we run `paradox(paradox)`?

- **Case A:** If `halts(paradox, paradox)` returns `True` (predicts termination), then `paradox(paradox)` enters an infinite loop — so it doesn't terminate. Contradiction.
- **Case B:** If `halts(paradox, paradox)` returns `False` (predicts non-termination), then `paradox(paradox)` returns immediately — so it does terminate. Contradiction.

**Conclusion:** Both outcomes contradict the prediction. Therefore, `halts(P, I)` cannot exist. No algorithm can solve the halting problem in the general case.

---

The core trick is **self-reference** — the program receives its own description as input, creating a logical loop analogous to the liar paradox ("This statement is false"). Turing showed this isn't a limitation of current technology but a fundamental boundary of computation itself.

Cantor's diagonalization argument proves the real numbers are uncountable by contradiction. Suppose the real numbers between 0 and 1 can be listed in a sequence, one for each natural number — meaning they are countable. Each number has an infinite decimal expansion, so the list looks like a table where the n-th row is the n-th number and the columns are its decimal digits. Now construct a new number by reading the diagonal: take the first digit of the first number, the second digit of the second number, and so on, and change each digit (for instance, add 1 modulo 10). The resulting number differs from the first number in its first digit, from the second number in its second digit, and in general from the n-th number in its n-th digit.

Because this new number differs from every number on the list in at least one position, it cannot appear anywhere in the list — yet it is a valid real number between 0 and 1. This contradicts the assumption that every such real number was already listed, so no such complete list can exist. Therefore the real numbers are strictly more numerous than the natural numbers: they are uncountable.

### For Loop

One way to fix the issue in [](#ex:infinite-loop) is to modify the condition of the while loop. (How?) However, we will learn a better fix as follows using the [`for` statement](https://docs.python.org/3/reference/compound_stmts.html#the-for-statement):

````python
for i in range(int(input(">")) + 1):
    print(i)
````

A simplified version of the syntax is:

```ebnf
for_stmt ::=  "for" target_list "in" starred_list ":" suite
```

A `for` loop is written as `for <variable> in <iterable>:` followed by an indented block; each iteration binds `<variable>` to the next element of the *iterable* (such as a list, tuple, range, or generator) and executes the indented statements under the colon, so the syntax is typically `for i in range(5):` on one line, then a consistently indented block (e.g., four spaces) for the body, as in `for i in range(5): print(i)`, where `print(i)` runs with `i` taking the values 0, 1, 2, 3, 4.

**How to print from 0 to 4?**

Let us visualize the execution of a simpler version of the for loop that prints from 1 up to 4:

````python
%%pytutor
for i in 1, 2, 3, 4:
    print(i)
````

- `i` is automatically assigned to each element in the sequence `1, 2, 3, 4` one-by-one from left to right.[^overwrite]
- After each assignment, the body `print(i)` is executed. 

[^overwrite]: If `i` is defined before the for loop, its value will be overwritten.

Unlike other languages such as C, Python's `for` loop uses the `in` keyword to
iterate over a collection of objects called an
[iterable](https://docs.python.org/3/glossary.html#term-iterable).

```c

    for (int i = 1; i <= 4; ++i) {
        printf("%d\n", i);
    }
    return 0;
}
```

One benefit of using iterables is that it is difficult to write an infinite
loop using the Python `for` loop.[^for-infinite-loop] Another reason is that
this allows Python to be more expressive and easy to read. For instance:

### While Loop vs For Loop

**How to decide whether to use while loop or for loop?**

::::{important} Definite vs indefinite loops

-  `for` is often used for a *definite loop* which has a definite number of iterations before execution.
- `while` is often used for an *indefinite loop* where the number of iterations is unknown before execution.
::::

It is always possible to replace a `for` loop by a `while` loop. Indeed, a `for` loop such as

```python
for i in range(5): print(i)
```

translates to the following code using a `while` loop:

````python
%%pytutor
iterable = range(5)
iterator = iter(iterable)
try:
    while True:
        i = next(iterator)
        print(i)
except StopIteration: pass
````

When executing the `for` loop, 
- the interpreter creates an iterator from the iterable, which tracks the next element to be assigned to the target variable.[^for-1]
- The loop terminates automatically when the iterator has no next element and raises an exception to signal the end of iteration.[^for-2]


[^for-1]: The `iter` function is used to create an iterator from an iterable, such as a range object. Each iteration of the `while` loop calls the `next` function on the iterator to retrieve the next value from the iterable.
[^for-2]: If there are no more next values, the iterator raises a `StopIteration` exception. The `except` block is used to catch this exception to prevent it from bubbling up further and crashing the program. The `try` block specifies the code to monitor for the exception to catch.

We can compare the speed of the `for` loop and `while` loop using the [cell magic `%%timeit`](https://ipython.readthedocs.io/en/stable/interactive/magics.html#magic-timeit) as follows:

````python
%%timeit
for i in range(5): cur_count = i
````

````python
%%timeit
i = 0
while i < 5:
    cur_count = i
    i += 1
````

### Breaking out of a loop

**Is the following an infinite loop?**

````python
while True:
    message = input("Input something please:")
    if message:
        break
print("You entered:", message)
````

The loop is terminated by the [`break` statement](https://docs.python.org/3/tutorial/controlflow.html#break-and-continue-statements-and-else-clauses-on-loops) when user input is non-empty.

**Why is the `break` statement useful?**

Recall the earlier `while` loop:

````python
while not input("Input something please:"):
    pass
````

This while loop is not useful because it does not store the user input.

**Is the `break` statement strictly necessary?**

No, as we can use [assignment expressions](https://peps.python.org/pep-0572/) starting with Python version 3.8.

More generally, we can avoid `break` statement by using *flags*, which are boolean variables for flow control:

````python
has_no_input = True
while has_no_input:
    message = input("Input something please:")
    if message:
        has_no_input = False
print("You entered:", message)
````

::::{note}

Using flags sometimes makes the program more readable, and we can use multiple flags for more complicated behavior.  
Flags are often named as `is_...`, `has_...`, etc.

### Else construct for a loop

The following program checks whether a number is composite, namely,  
- a positive integer that is
- a product of two strictly smaller positive integers.

````python
@interact(num="1")
def check_composite(num):
    if num.isdigit():
        num = int(num)
        for divisor in range(2, num):  # why starts from 2 instead of 1
            if not num % divisor:
                print("It is composite.")
                break  # where will this go?
        else:
            print("It is not composite.")  # how to get here?
    else:
        print("Not a positive integer.")  # how to get here?
````

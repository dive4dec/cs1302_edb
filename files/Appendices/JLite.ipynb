[JupyterLite](https://jupyterlite.readthedocs.io/) is a version of JupyterLab that runs **entirely in your browser** — no server-side Python kernel is required. In the course environment, JLite provides a self-contained computing environment where you can run Python code, interact with AI assistants, and view course materials — all without a backend server.

You can access the JLite site at: **https://dive4dec.github.io/cs1302_edb/**

## How JupyterLite Works

Traditional Jupyter requires a server running Python kernels. JupyterLite replaces this architecture with **WebAssembly (Wasm)** — Python code is compiled to Wasm and executes directly in the browser:

```
┌─────────────────────────────────────────────┐
│  Your Browser                                │
│  ┌─────────────┐  ┌──────────────────────┐  │
│  │  JupyterLab  │  │  xeus-python (Wasm)  │  │
│  │  Frontend    │◄─┤  Python 3.13 Kernel  │  │
│  │              │  │  numpy, pandas,      │  │
│  │  AI Chat     │  │  matplotlib, etc.    │  │
│  └─────────────┘  └──────────────────────┘  │
└─────────────────────────────────────────────┘
```

Key components in the course JLite site:

| Component | Role |
|-----------|------|
| **xeus-python** | In-browser Python kernel compiled to WebAssembly (Python 3.13) |
| **jupyterlite-ai** | AI chat and code completion extension (v0.18.0) |
| **jupyterlab-myst** | MyST markdown rendering in notebooks |
| **divewidgets** | Custom interactive widgets for the course |
| **jupytext** | Notebook-to-markdown conversion support |

## In-Browser Python Kernel (xeus-python)

The course JLite site uses the **xeus-python** kernel — a Python implementation built on the [xeus](https://github.com/jupyter-xeus/xeus) messaging protocol, compiled to WebAssembly via Emscripten.

### Available Packages

The kernel includes these pre-packaged libraries (compiled to Wasm):

| Package | Version | Description |
|---------|---------|-------------|
| **Python** | 3.13.1 | Core Python runtime |
| **numpy** | 2.4.6 | Numerical computing |
| **pandas** | 3.0.3 | DataFrames and data analysis |
| **matplotlib** | 3.10.9 | Plotting and visualization |
| **ipywidgets** | 8.1.8 | Interactive widgets |
| **jupyterlab-myst** | 2.3.1 | MyST markdown in notebooks |

| **scipy** | included | Scientific computing |
| **IPython** | 9.9.0 | Enhanced REPL |
| **pillow** | 12.2.0 | Image processing |

### Limitations of In-Browser Python

Since everything runs in WebAssembly:

- **No system access** — you cannot run shell commands, install packages, or access the filesystem beyond the browser's virtual storage
- **No network requests** — cannot fetch data from URLs or connect to external APIs
- **No multithreading** — CPU-intensive tasks are limited to a single thread
- **Memory limits** — constrained by browser memory allocation (typically ~2GB)
- **No native extensions** — packages requiring C/C++ compilation (like some ML libraries) are not available

Despite these limitations, the kernel is fully functional for the course's computational exercises, data analysis, and visualization tasks.

## JupyterLite AI Extension

The [`@jupyterlite/ai`](https://github.com/jupyterlite/ai) extension (v0.18.0) provides AI-powered features directly in the JupyterLite interface. All AI processing happens via **cloud API calls** (not locally in the browser), so the model's capabilities depend on your configured provider.

### Features

- **AI Chat Panel** — conversational AI assistant for code help, explanations, and debugging
- **Code Completion** — inline code suggestions as you type
- **Code Execution** — AI can execute code in the kernel and view results
- **Notebook Operations** — AI can create, edit, and run cells
- **Context Awareness** — AI sees your current notebook and file content
- **Tool Use** — AI can use tools for file management and notebook operations
- **MCP Support** — Model Context Protocol server integration

### Configuring AI Providers

The JupyterLite AI extension supports multiple providers via the Vercel AI SDK:

| Provider | Package | Models |
|----------|---------|--------|
| **Anthropic** | `@ai-sdk/anthropic` | Claude Sonnet, Opus, Haiku |
| **OpenAI** | `@ai-sdk/openai` | GPT-4, GPT-3.5, o-series |
| **Google** | `@ai-sdk/google` | Gemini Pro, Flash, Ultra |
| **Mistral** | `@ai-sdk/mistral` | Mistral Large, Medium, Small |
| **OpenAI-compatible** | `@ai-sdk/openai-compatible` | Any OpenAI-compatible API (OpenRouter, LM Studio, etc.) |

To configure your AI provider:

1. Open the **Settings** menu in JupyterLite
2. Navigate to **JupyterLite AI Settings**
3. Add a provider configuration with your API key
4. Set the **Default Chat Provider** and optionally a separate **Completion Provider**

Example provider configuration:

```json
{
  "id": "my-anthropic",
  "name": "Anthropic",
  "provider": "anthropic",
  "model": "claude-sonnet-4-20250514",
  "apiKey": "your-api-key-here",
  "parameters": {
    "temperature": 0.7,
    "maxOutputTokens": 4096
  }
}
```

### Security

- API keys are stored using the **secrets manager** by default (encrypted in browser storage)
- Set `"useSecretsManager": false` to store keys in plain text settings (not recommended)
- The extension uses the **Vercel AI SDK** which handles streaming and error recovery

### AI Capabilities in JupyterLite

The AI extension has several configurable capabilities:

| Setting | Default | Description |
|---------|---------|-------------|
| `toolsEnabled` | `true` | AI can use tools (notebook/file operations) |
| `codeExecution` | `false` | AI can execute code in the kernel |
| `contextAwareness` | `true` | AI sees current notebook context |
| `autoTitle` | `false` | Auto-generate chat titles |
| `showTokenUsage` | `false` | Display token usage stats |
| `showCellDiff` | `true` | Show diffs when AI modifies cells |

The AI can use the following JupyterLab commands (with user approval):

- `notebook:run-cell` — execute cells
- `notebook:run-all-above` / `notebook:run-all-below` — run cell ranges
- `jupyterlab-ai-commands:execute-in-kernel` — direct kernel execution

## Using JLite for Course Work

### Opening Notebooks

The JLite site serves course materials from the `release/` directory. You can:

1. **Browse notebooks** — navigate through Lab and Lecture folders
2. **Execute code** — run cells directly in the browser with the xeus-python kernel
3. **Create new notebooks** — make copies and modify them for exercises
4. **Save changes** — modifications persist in your browser's local storage

### AI-Assisted Learning

Use the JupyterLite AI to:

- **Ask questions** about course concepts — the AI sees your notebook context
- **Debug code** — paste error messages and get explanations
- **Generate code** — ask the AI to write functions, plots, or data analysis
- **Explain concepts** — get detailed explanations with examples
- **Complete exercises** — use AI as a study companion (not a substitute)

Example prompts to try:

```
Explain why this cell is producing an error
Write a function that calculates the BMI given weight and height
Create a matplotlib bar chart showing these categories
```

### Limitations to Keep in Mind

- **No file I/O** — cannot read/write files beyond the virtual filesystem
- **No package installation** — all packages are pre-bundled
- **No external APIs** — cannot call web services or fetch remote data
- **Session persistence** — your work is saved in browser storage, but clearing cookies/ cache may lose unsaved changes
- **Performance** — heavy computations may be slower than a server kernel

## Architecture Details

### Build Configuration

The JLite site is built using these configurations:

| File | Purpose |
|------|---------|
| `jupyterlite/jupyter_lite_config.json` | Points `LiteBuildConfig.contents` to `../release` |
| `jupyterlite/environment.yml` | Kernel packages for xeus-python |
| `jupyterlite/dep/environment.yml` | Build environment (Python 3.14 + jupyterlite-core) |
| `jupyterlite/dep/requirements.txt` | Build-time pip dependencies |

The build process (via `jupyterlite/Makefile`):

```
make env    → create conda environment from dep/environment.yml
make build  → jupyter lite build --output-dir _output
make serve  → python -m http.server --directory _output (local preview)
make publish → ghp-import to release/ submodule
```

## Glossary

:::::{admonition}
:class: dropdown

::::{glossary}

JupyterLite
: A serverless version of JupyterLab that runs entirely in the browser using WebAssembly. No backend server is needed — kernels execute as compiled Wasm modules.

xeus-python
: A Python kernel implementation built on the xeus messaging protocol, compiled to WebAssembly. Provides a full Python 3.13 environment in the browser.

WebAssembly (Wasm)
: A binary instruction format that enables near-native performance in web browsers. Used to run Python code directly in the browser without a server.

JupyterLite AI
: The `@jupyterlite/ai` extension that provides AI chat and code completion in JupyterLite. Uses the Vercel AI SDK to communicate with cloud AI providers.

Vercel AI SDK
: A JavaScript SDK for building AI-powered applications. In JupyterLite AI, it handles streaming, provider abstraction, and error recovery for AI API calls.

MCP (Model Context Protocol)
: A protocol that allows AI models to access external tools and data sources. Supported by JupyterLite AI for extended capabilities.

Emscripten
: A compiler toolchain that compiles C/C++ code to WebAssembly. Used to compile the xeus-python kernel and its dependencies for browser execution.
